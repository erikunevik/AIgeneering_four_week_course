```python
from pathlib import Path

import pandas as pd

data_dir = Path("../data")
files = sorted(data_dir.glob("ads*.txt"))

job_ads = []

for path in files:
    text = path.read_text(encoding="utf-8")
    job_ads.append(
        {
            
            "filename":path.name,
            "text": text,
         
         }
        
    )
    

df = pd.DataFrame(job_ads)
print(df)   
```

##b

```python
from google import genai

client = genai.Client()

def summarize_ads(df, model="gemini-2.5-flash"):
    def get_summary(ad_text):
        response = client.models.generate_content(
            model=model,
            contents=f"Summarize this job ad:{ad_text}"
        )
        return response.text
    
    df["summary"] = df["text"].apply(get_summary)
    return df

# Använd den
df = summarize_ads(df)

print(df.loc[0, "summary"])
```

```python
def converttxt_tomd(df):

    dir_folder = Path("../data")

    for row in df.iterrows():
        md_filename = row["filename"].replace(".txt", ".md")
        md_path = dir_folder / md_filename
        
        content = f"# {row['filename']}\n\n"
        content += f"##-----------------Summarized ad----------------------\n\n"
        content += f"####  {row['summary']}  \n\n"
        content += f"##-----------------Main text----------------------\n\n"
        content += f"#### {row['text']}\n\n"
        content += f"---------------------------------------\n\n"
        
        
        md_path.write_text(content, encoding="utf-8")
        
    
```

```python
from pathlib import Path

import pandas as pd

data_dir = Path("exams")
files = sorted(data_dir.glob("tex*.txt"))

test_list = []

for path in files:
    text = path.read_text(encoding="utf-8")
    test_list.append(
        {
            
            "filename":path.name,
            "text": text,
         
         }
        
    )
    
```

```python
test_list
```

```python
from pydantic import BaseModel
from google import genai

class Grade(BaseModel):
    proposed_grade: str
    motivation: str
    improvments: str  

client = genai.Client()

grades: list[Grade] = []   

for item in test_list:
    text = item["text"]   

    response = client.models.generate_content(
        model="gemini-2.0-flash",
        contents=f"""
        You are a teacher. Read this text and respond ONLY as valid JSON
        that matches this schema: Grade(proposed_grade, motivation, improvments).

        Text:
        {text}
        """,
        config={
            "response_mime_type": "application/json",
            "response_schema": Grade,  # EN Grade per text
        },
    )

    grade_obj = Grade.model_validate_json(response.text)
    grades.append(grade_obj)
```
