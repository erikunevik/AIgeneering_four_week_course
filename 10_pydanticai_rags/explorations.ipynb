## Pdf to text

```python
from pypdf import PdfReader

reader = PdfReader("data/Dwarf_Hotot.pdf")

reader
```

```python
all_text = ""

for page in reader.pages:
    all_text += page.extract_text()
    
all_text
```

```python
len(all_text)
```
